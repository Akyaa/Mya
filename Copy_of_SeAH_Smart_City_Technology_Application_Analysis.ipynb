```python

import pandas as pd
import re
from collections import Counter

# =====================================================
# URBAN APPLICATION DOMAIN DICTIONARY
# =====================================================

APPLICATION_DOMAINS = {
    "Transportation and Mobility": [
        r"\btransportation\b",
        r"\btransport\b",
        r"\btraffic\b",
        r"\bmobility\b",
        r"\bsmart parking\b",
        r"\bpublic transit\b",
        r"\broad networks?\b",
        r"\bride.hailing\b",
    ],

    "Energy": [
        r"\benergy management\b",
        r"\brenewable energy\b",
        r"\bsolar energy\b",
        r"\bsolar photovoltaic\b",
        r"\belectricity\b",
        r"\bsmart grids?\b",
        r"\bmicrogrids?\b",
        r"\bpower distribution\b",
    ],

    "Housing and Buildings": [
        r"\bhousing\b",
        r"\bresidential buildings?\b",
        r"\bsmart buildings?\b",
        r"\bbuilding automation\b",
        r"\bconstruction\b",
        r"\bfacility management\b",
        r"\breal estate\b",
    ],

    "Water Management": [
        r"\bwater supply\b",
        r"\bwater management\b",
        r"\bwater quality\b",
        r"\bwastewater\b",
        r"\bdrainage\b",
        r"\bwater distribution\b",
        r"\bsmart water\b",
        r"\bflood\b",
    ],

    "Waste Management": [
        r"\bwaste management\b",
        r"\bsolid waste\b",
        r"\bwaste collection\b",
        r"\brecycling\b",
        r"\bwaste disposal\b",
        r"\billegal dumping\b",
    ],

    "Environment and Sustainability": [
        r"\bair quality\b",
        r"\bair pollution\b",
        r"\bclimate change\b",
        r"\burban heat\b",
        r"\bheat island\b",
        r"\benvironmental monitoring\b",
        r"\benvironmental sustainability\b",
        r"\burgan ecology\b",
    ],

    "Governance and Public Services": [
        r"\be.government\b",
        r"\be.governance\b",
        r"\bsmart governance\b",
        r"\burban governance\b",
        r"\bmunicipal services?\b",
        r"\bpublic services?\b",
        r"\bcitizen participation\b",
        r"\bpublic administration\b",
    ],
}

# =====================================================
# CLASSIFY PUBLICATIONS
# =====================================================

domain_results = []

for index, row in df.iterrows():

    text = " ".join([
        str(row.get("Title", "") or ""),
        str(row.get("Abstract", "") or ""),
        str(row.get("Author Keywords", "") or "")
    ])

    detected_domains = []

    for domain, patterns in APPLICATION_DOMAINS.items():
        if any(
            re.search(pattern, text, flags=re.IGNORECASE)
            for pattern in patterns
        ):
            detected_domains.append(domain)

    domain_results.append({
        "Record Number": index + 1,
        "Title": row["Title"],
        "Year": row["Year"],
        "Detected Application Domains":
            "; ".join(detected_domains),
        "Number of Domains": len(detected_domains)
    })

domain_df = pd.DataFrame(domain_results)

print("URBAN APPLICATION CLASSIFICATION")
print("=" * 55)
print("Publications examined:", len(domain_df))
print(
    "Publications with domain matches:",
    (domain_df["Number of Domains"] > 0).sum()
)
print(
    "Publications without matches:",
    (domain_df["Number of Domains"] == 0).sum()
)

display(domain_df.head())
```

```python

import re
import pandas as pd
from collections import Counter

# Correct the spelling error in the environment dictionary
APPLICATION_DOMAINS["Environment and Sustainability"] = [
    pattern.replace("urgan ecology", "urban ecology")
    for pattern in APPLICATION_DOMAINS["Environment and Sustainability"]
]

# Collect matched terms for each domain
term_counts = {
    domain: Counter()
    for domain in APPLICATION_DOMAINS
}

for _, row in df.iterrows():

    text = " ".join(
        str(row[field]) if pd.notna(row[field]) else ""
        for field in ["Title", "Abstract", "Author Keywords"]
    )

    for domain, patterns in APPLICATION_DOMAINS.items():

        publication_terms = set()

        for pattern in patterns:
            for match in re.finditer(
                pattern, text, flags=re.IGNORECASE
            ):
                publication_terms.add(match.group(0).lower())

        term_counts[domain].update(publication_terms)

print("MOST FREQUENT MATCHED TERMS BY APPLICATION DOMAIN")
print("=" * 65)

for domain, counts in term_counts.items():
    print(f"\n{domain}")
    print("-" * 50)

    for term, count in counts.most_common(10):
        print(f"{term}: {count}")
```

```python

import pandas as pd

# Terms that can produce overly broad classifications
BROAD_TERMS = {
    "construction",
    "climate change",
    "transport",
    "housing",
    "public service",
    "public services"
}

review = application_review.copy()

def identify_broad_terms(value):
    terms = {
        term.strip().lower()
        for term in str(value).split(";")
    }
    return "; ".join(sorted(terms & BROAD_TERMS))

review["Broad Terms Flagged"] = (
    review["Matched Terms"].apply(identify_broad_terms)
)

# Identify publications assigned to multiple domains
domain_frequency = review.groupby(
    "Record Number"
)["Candidate Application Domain"].nunique()

review["Multiple Domains"] = (
    review["Record Number"].map(domain_frequency) > 1
)

# Prioritize broad terms and overlapping classifications
priority_review = review[
    (review["Broad Terms Flagged"] != "") |
    (review["Multiple Domains"])
].copy()

priority_review = priority_review.sort_values(
    ["Record Number", "Candidate Application Domain"]
)

print("HIGH-PRIORITY DOMAIN VERIFICATION")
print("=" * 55)
print("Total candidate relationships:", len(review))
print("High-priority relationships:", len(priority_review))
print(
    "Unique high-priority publications:",
    priority_review["Record Number"].nunique()
)

print("\nHigh-priority relationships by domain:")
display(
    priority_review[
        "Candidate Application Domain"
    ].value_counts().rename_axis(
        "Application Domain"
    ).reset_index(name="Review Tasks")
)

# Save a separate workbook
priority_file = "/content/priority_domain_verification.xlsx"

priority_review.to_excel(
    priority_file,
    index=False
)

print("\nWorkbook created:", priority_file)
```

```python

import pandas as pd
import re

BROAD_PATTERNS = {
    "transport": r"\btransport\b",
    "construction": r"\bconstruction\b",
    "climate change": r"\bclimate change\b",
    "housing": r"\bhousing\b",
    "public services": r"\bpublic services\b",
    "public service": r"\bpublic service\b"
}

results = []

for term, pattern in BROAD_PATTERNS.items():

    counts = {
        "Title": 0,
        "Author Keywords": 0,
        "Abstract Only": 0
    }

    for _, row in df.iterrows():

        title = str(
            row["Title"] if pd.notna(row["Title"]) else ""
        )

        keywords = str(
            row["Author Keywords"]
            if pd.notna(row["Author Keywords"]) else ""
        )

        abstract = str(
            row["Abstract"] if pd.notna(row["Abstract"]) else ""
        )

        in_title = bool(re.search(pattern, title, re.I))
        in_keywords = bool(re.search(pattern, keywords, re.I))
        in_abstract = bool(re.search(pattern, abstract, re.I))

        if in_title:
            counts["Title"] += 1

        if in_keywords:
            counts["Author Keywords"] += 1

        if in_abstract and not in_title and not in_keywords:
            counts["Abstract Only"] += 1

    results.append({
        "Search Term": term,
        **counts
    })

term_location_summary = pd.DataFrame(results)

print("BROAD TERM LOCATION ANALYSIS")
print("=" * 55)

display(term_location_summary)
```

```python

import pandas as pd
import re
from google.colab import files

review_records = []

for _, row in priority_review.iterrows():

    record_number = int(row["Record Number"])
    source = df.iloc[record_number - 1]

    title = str(source["Title"]) if pd.notna(source["Title"]) else ""
    keywords = (
        str(source["Author Keywords"])
        if pd.notna(source["Author Keywords"])
        else ""
    )
    abstract = (
        str(source["Abstract"])
        if pd.notna(source["Abstract"])
        else ""
    )

    for term, pattern in BROAD_PATTERNS.items():

        if not re.search(
            pattern, str(row["Broad Terms Flagged"]), re.I
        ):
            continue

        in_title = bool(re.search(pattern, title, re.I))
        in_keywords = bool(re.search(pattern, keywords, re.I))
        in_abstract = bool(re.search(pattern, abstract, re.I))

        if in_abstract and not in_title and not in_keywords:
            review_records.append({
                "Record Number": record_number,
                "EID": source.get("EID", ""),
                "Year": source["Year"],
                "Title": title,
                "Author Keywords": keywords,
                "Abstract": abstract,
                "Candidate Application Domain":
                    row["Candidate Application Domain"],
                "Broad Term": term,
                "Decision": "",
                "Final Application Domain": "",
                "Reviewer Notes": ""
            })

abstract_review = pd.DataFrame(review_records)

# Prevent duplicated review tasks
abstract_review = abstract_review.drop_duplicates(
    subset=[
        "Record Number",
        "Candidate Application Domain",
        "Broad Term"
    ]
)

print("ABSTRACT-ONLY VERIFICATION")
print("=" * 55)
print("Review tasks:", len(abstract_review))
print(
    "Unique publications:",
    abstract_review["Record Number"].nunique()
)

display(
    abstract_review[
        "Candidate Application Domain"
    ].value_counts()
)

output_path = "/content/abstract_only_domain_review.xlsx"

abstract_review.to_excel(output_path, index=False)

files.download(output_path)
```

```python

import pandas as pd

print("SECTION IV.A: PUBLICATION TRENDS AND RESEARCH PERFORMANCE")
print("=" * 70)

# Confirm dataset
print("\nDATASET OVERVIEW")
print("Total publications:", len(df))
print("Available columns:")
print(df.columns.tolist())

# Prepare publication years
years = pd.to_numeric(df["Year"], errors="coerce")

print("\nPUBLICATION YEAR VALIDATION")
print("Earliest year:", years.min())
print("Latest year:", years.max())
print("Missing or invalid years:", years.isna().sum())

# Calculate annual publication counts
annual_publications = (
    years.value_counts()
    .sort_index()
    .rename_axis("Year")
    .reset_index(name="Publications")
)

annual_publications["Year"] = (
    annual_publications["Year"].astype(int)
)

# Calculate annual growth
annual_publications["Annual Growth (%)"] = (
    annual_publications["Publications"]
    .pct_change() * 100
)

annual_publications["Cumulative Publications"] = (
    annual_publications["Publications"].cumsum()
)

print("\nANNUAL PUBLICATION PERFORMANCE")
display(annual_publications.round(2))

print("\nPUBLICATION SUMMARY")
print("Total:", annual_publications["Publications"].sum())

highest = annual_publications.loc[
    annual_publications["Publications"].idxmax()
]

print(
    "Peak publication year:",
    int(highest["Year"])
)

print(
    "Publications in peak year:",
    int(highest["Publications"])
)

# Check citation availability
print("\nCITATION DATA")

if "Cited by" in df.columns:
    citations = pd.to_numeric(
        df["Cited by"], errors="coerce"
    )

    print("Records with numeric citation data:",
          citations.notna().sum())

    print("Total recorded citations:",
          citations.sum())

    print("Mean citations per publication:",
          round(citations.mean(), 2))

    print("Median citations:",
          citations.median())

    print("Maximum citations:",
          citations.max())

    print("Missing citation values:",
          citations.isna().sum())
else:
    print("Citation column not found.")

# Inspect publication document types
print("\nDOCUMENT TYPES")

if "Document Type" in df.columns:
    print(
        df["Document Type"]
        .value_counts(dropna=False)
        .to_string()
    )
else:
    print("Document Type column not found.")
```

```python

import pandas as pd

print("ANNUAL CITATION PERFORMANCE")
print("=" * 65)

# Prepare year and citation data
citation_df = df[["Year", "Cited by"]].copy()

citation_df["Year"] = pd.to_numeric(
    citation_df["Year"],
    errors="coerce"
)

citation_df["Citations"] = pd.to_numeric(
    citation_df["Cited by"],
    errors="coerce"
)

# Calculate annual citation statistics
annual_citations = (
    citation_df.groupby("Year")
    .agg(
        Publications=("Citations", "size"),
        Total_Citations=("Citations", "sum"),
        Mean_Citations=("Citations", "mean"),
        Median_Citations=("Citations", "median"),
        Maximum_Citations=("Citations", "max")
    )
    .reset_index()
)

annual_citations["Year"] = (
    annual_citations["Year"].astype(int)
)

# Round for presentation
annual_citations["Mean_Citations"] = (
    annual_citations["Mean_Citations"].round(2)
)

annual_citations["Median_Citations"] = (
    annual_citations["Median_Citations"].round(2)
)

print("\nANNUAL CITATION PERFORMANCE TABLE")
display(annual_citations)

print("\nCITATION SUMMARY")

print(
    "Total citations:",
    int(annual_citations["Total_Citations"].sum())
)

most_cited_year = annual_citations.loc[
    annual_citations["Total_Citations"].idxmax()
]

highest_average_year = annual_citations.loc[
    annual_citations["Mean_Citations"].idxmax()
]

print(
    "Year with highest total citations:",
    int(most_cited_year["Year"])
)

print(
    "Highest annual citation total:",
    int(most_cited_year["Total_Citations"])
)

print(
    "Year with highest mean citations:",
    int(highest_average_year["Year"])
)

print(
    "Highest mean citations per publication:",
    highest_average_year["Mean_Citations"]
)
```
